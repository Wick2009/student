```python
%%javascript

// Hack 2 – Count Repeats
let items = ["cat", "dog", "cat", "bird", "bird", "bird"];

// Count how many times each item appears
let counts = {};
for (let it of items) {
  counts[it] = (counts[it] || 0) + 1;
}

console.log("--- Hack 2: Count Repeats ---");
console.log(`Items: ${items}`);
console.log("Counts:", counts);
```

### Hack 1 – Add Up Numbers

Make a list of numbers. Write code to:
1. Find the total sum.
2. Find the average.

```python
%%javascript

// Hack 1 – Add Up Numbers
let numbers = [4, 7, 1, 9, 6, 7, 10];

// Sum the numbers
let total = numbers.reduce((acc, n) => acc + n, 0);
let average = total / numbers.length;

console.log("--- Hack 1: Add Up Numbers ---");
console.log(`Numbers: ${numbers}`);
console.log(`Total: ${total}`);
console.log(`Average: ${average}`);
```
