### Hack 1: How Much Time?

#### Objective: write the time complexity of the algorithm below using Big-O notation.
(don't worry about special cases such as n = 1 or n = 0).

```python
n = int(input()) # remember what O(n) means? This is a good way of visualizing n.

for i in range(n):
    print(i)

# Time complexity: O(n) — the loop runs once per element, so it grows linearly with n
print('Time complexity: O(n)')
```

### Hack 2: Your Turn!

#### Objective: <strong>write</strong> an algorithm with O(n^2) time complexity.

```python
n = int(input())

# O(n^2) example: nested loops that visit every pair (i, j)
ops = 0
for i in range(n):
    for j in range(n):
        # constant work per pair
        ops += 1

print('Total pair operations:', ops)
print('Time complexity: O(n^2)')
```

### Hack 3: Gotta Go Fast!

#### Objective: Optimize this algorithm so that it has a lower time complexity <strong>without modifying the outer loop</strong>

```python
import math
n = int(input())
count = 0

# Precompute the number of inner iterations (this is independent of i)
inner_iterations = math.ceil(math.sqrt(n) * 2)

for i in range(n):
    # Instead of running the inner loop repeatedly, add the fixed amount each time
    count += inner_iterations

print(count)

# Explanation:
# The original nested structure performed inner_iterations for each of the n outer iterations,
# giving O(n * inner_iterations). Since inner_iterations depends on n but not on i, the
# original complexity was O(n * sqrt(n)) = O(n^{1.5}). By precomputing inner_iterations and
# adding it once per outer iteration, each outer iteration does O(1) work, so the overall
# complexity becomes O(n).
print('Optimized time complexity: O(n) — original was O(n * sqrt(n))')
```

### Hack 4: Extra Challenge 

#### Objective: Write an algorithm that does <strong>NOT</strong> have a time complexity of O(1), O(n), or O(n^x) and identify the time complexity
##### (I will not accept O(n^3) or some other power, it needs to be more complex.)

```python
n = int(input('Enter n (small, e.g., 10): '))

# Exponential example: generate all subsets of a set of size n
# This enumerates 2^n subsets, so the algorithm has time complexity O(2^n)
if n > 20:
    print('n too large for this demo; try n <= 20')
else:
    subsets = []
    for mask in range(1 << n):  # 0..2^n - 1
        subset = []
        for i in range(n):
            if mask & (1 << i):
                subset.append(i)
        subsets.append(subset)

    print('Number of subsets generated:', len(subsets))
    print('Example first 5 subsets:', subsets[:5])
    print('Time complexity: O(2^n) — exponential, not polynomial (not O(n^x))')
```
