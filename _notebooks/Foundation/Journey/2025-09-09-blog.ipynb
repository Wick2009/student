# My Journey: Setting Up VS Code and Learning CSP on Windows
Switching from a MacBook to a Windows computer for my CSP class was a big change! Here’s how I set up VS Code, learned new tools, and overcame challenges to get everything working.

---
## 🖥️ Installing VS Code and Extensions
- Downloaded VS Code from the official site and installed it.
- Made sure to add VS Code to PATH and enable context menu options for easy access.
- Installed key extensions:
    - Python (Microsoft) for coding and debugging
    - Jupyter (Microsoft) for working with notebooks
    - GitLens for advanced Git features
- Verified installations using the terminal:
    - `python --version` to check Python
    - `git --version` to check Git
    - Configured Git with my name and email.

---
## 📂 Managing Repositories and Personalization
- Cloned the main pages repository into my opencs folder.
- Experimented with themes and made personal changes to my site.
- Created my own student repository for personal work.

---
## 🐍 Setting Up a Virtual Environment
- Used a setup script to create a Python virtual environment:
    - `./scripts/venv.sh`
- Activated the environment before running code:
    - `source venv/bin/activate`
---
## 🤝 Collaborating with My Team
- Created and forked shared repositories for team projects.

---
## 📓 Working with Jupyter Notebooks
- Located and opened notebook files in VS Code.
- Used Developer Tools to clear the console and ran cells with the play button.
- Edited code and outputs to personalize my notebooks (like adding my own jokes!).
- Committed changes with outputs using Git commands.

---
## ✅ Other Accomplishments
- Verified notebook outputs appeared correctly on my site.
- Ensured Python code ran smoothly in the virtual environment.
- Managed multiple repositories and customized project settings.
- Got VS Code working perfectly on Windows!
